# 线性回归，闭式解
Scikit-learn 有[线性回归模型](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html#sklearn.linear_model.LinearRegression)，该模型实现了闭式线性回归。

让我们使用早期实验中的数据——一套 1000 平方英尺的房子售价为 30 万美元，一套 2000 平方英尺的房子售价为 50 万美元。
|面积（1000 平方英尺）|价格（千美元）|
| ------| ------|
|1|300|
|2|500|


### 加载数据集

```python
X_train = np.array([1.0, 2.0])   #features
y_train = np.array([300, 500])   #target value
```

### 创建并拟合模型
以下代码使用 scikit-learn 进行回归。第一步创建一个回归对象。第二步使用与该对象相关联的方法之一“fit”。这执行回归操作，将参数拟合到输入数据。该工具包期望一个二维的 X 矩阵。

```python
linear_model = LinearRegression()
#X must be a 2-D Matrix
linear_model.fit(X_train.reshape(-1, 1), y_train) 
```

### 视图参数
在 scikit-learn 中，参数 $\mathbf{w}$ 和 $\mathbf{b}$ 分别被称为“系数”和“截距”。

```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
print(f"'manual' prediction: f_wb = wx+b : {1200*w + b}")
```

### 进行预测
调用`predict`函数可生成预测结果。

```python
y_pred = linear_model.predict(X_train.reshape(-1, 1))

print("Prediction on training set:", y_pred)

X_test = np.array([[1200]])
print(f"Prediction for 1200 sqft house: ${linear_model.predict(X_test)[0]:0.2f}")
```

## 第二个示例
第二个示例来自早期的一个具有多个特征的实验室。最终的参数值和预测结果与该实验室未归一化的“长期运行”结果非常接近。那次未归一化的运行需要数小时才能产生结果，而这次几乎是瞬间完成。封闭形式解在较小的数据集上表现良好，例如这些数据集，但在较大的数据集上可能计算量很大。
>封闭形式解不需要归一化。

```python
# load the dataset
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train) 
```

```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
```

```python
print(f"Prediction on training set:\n {linear_model.predict(X_train)[:4]}" )
print(f"prediction using w,b:\n {(X_train @ w + b)[:4]}")
print(f"Target values \n {y_train[:4]}")

x_house = np.array([1200, 3,1, 40]).reshape(-1,4)
x_house_predict = linear_model.predict(x_house)[0]
print(f" predicted price of a house with 1200 sqft, 3 bedrooms, 1 floor, 40 years old = ${x_house_predict*1000:0.2f}")
```
